```python
# prompt: antes de copiar la nueva tabla eliminar la ultima fila de la tabla donde se encuentra el "TOTAL". ELIMINAR TODA ESA FILA. DAME TODO EL CODIGO

import pandas as pd
from docx import Document
from datetime import datetime
from docx.shared import Pt
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT
from docx.oxml.ns import nsmap, qn
from docx.oxml import OxmlElement
from docx.oxml.ns import qn
from docx.shared import RGBColor  # Para cambiar el color de las letras

# Rutas de los archivos
excel_path = "/content/08ENE-14ENE.xlsx"  # Reemplaza con la ruta correcta de tu archivo Excel
word_path = "/content/Suppler_Letter_01ENE-07ENE.docx"  # Reemplaza con la ruta correcta de tu archivo Word
output_path = "/content/Suppler_Letter_{fecha}.docx"  # Formato para guardar el nuevo archivo Word

# Leer la tabla desde el Excel y reemplazar NaN con valores vacíos
df = pd.read_excel(excel_path)
df = df.fillna("")  # Reemplazar NaN con cadenas vacías

# Convertir la columna 'Invoice' a tipo entero
if 'Invoice' in df.columns:
    df['Invoice'] = pd.to_numeric(df['Invoice'], errors='coerce').astype('Int64')

# Convertir la columna 'Amount' a tipo moneda y redondear currency
if 'Currency' in df.columns:
    df['Currency'] = df['Currency'].round(0).astype(int)

if 'Amount' in df.columns:
    df['Amount'] = df['Amount'].apply(lambda x: "${:,.2f}".format(x) if pd.notnull(x) else "")

# Abrir el documento Word existente
doc = Document(word_path)

# Eliminar la tabla anterior si existe
if doc.tables:
    print("Eliminando la tabla existente...")
    tbl = doc.tables[0]
    tbl._element.getparent().remove(tbl._element)

# Eliminar la oración si existe
target_text = "Espero que esto te sea útil. Si necesitas algún ajuste, házmelo saber."
for paragraph in doc.paragraphs:
    if target_text in paragraph.text.strip():
        print("Eliminando la oración...")
        p = paragraph._element
        p.getparent().remove(p)
        break

# Crear la nueva tabla
table = doc.add_table(rows=1, cols=len(df.columns))

# Aplicar bordes visibles a la tabla
def set_table_borders(table):
    tbl = table._element
    tbl_pr = tbl.xpath("./w:tblPr")[0]
    tbl_borders = OxmlElement('w:tblBorders')
    for border_name in ['top', 'left', 'bottom', 'right', 'insideH', 'insideV']:
        border = OxmlElement(f'w:{border_name}')
        border.set(qn('w:val'), 'single')
        border.set(qn('w:sz'), '12')  # Grosor del borde
        border.set(qn('w:space'), '0')
        border.set(qn('w:color'), '000000')  # Color negro
        tbl_borders.append(border)
    tbl_pr.append(tbl_borders)

set_table_borders(table)

# Añadir encabezados y datos alineados
hdr_cells = table.rows[0].cells
for i, col_name in enumerate(df.columns):
    hdr_cells[i].text = str(col_name)

    # Cambiar fondo azul y letras blancas en el encabezado
    hdr_cells[i].paragraphs[0].runs[0].bold = True
    hdr_cells[i].paragraphs[0].alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    hdr_cells[i].paragraphs[0].runs[0].font.color.rgb = RGBColor(255, 255, 255)  # Letras blancas
    hdr_cells[i].paragraphs[0].runs[0].font.size = Pt(11)

    # Cambiar fondo del encabezado a azul
    shading_elm = OxmlElement('w:shd')  # Crear un elemento de sombreado
    shading_elm.set(qn('w:fill'), '0000FF')  # Fondo azul
    hdr_cells[i]._element.get_or_add_tcPr().append(shading_elm)

# Añadir los datos de la tabla
for index, row in df.iterrows():
    # Verificar si la fila contiene solo celdas vacías o <NA>
    if row.isnull().all() or (row == "<NA>").all():
        continue  # Saltar la fila si está vacía o contiene solo <NA>

    row_cells = table.add_row().cells
    for i, value in enumerate(row):
        if isinstance(value, (int, float)):
            # Separador de miles para cantidades
            row_cells[i].text = "{:,}".format(value) if pd.notnull(value) else ""
        else:
            row_cells[i].text = str(value)
        row_cells[i].paragraphs[0].alignment = WD_PARAGRAPH_ALIGNMENT.CENTER

# Aplicar negrita a la última fila (totales) solo si la tabla no está vacía
if table.rows:
    last_row_cells = table.rows[-1].cells
    for cell in last_row_cells:
        for paragraph in cell.paragraphs:
            for run in paragraph.runs:
                run.bold = True

# Eliminar bordes inferiores de las celdas vacías
for row in table.rows:
    for cell in row.cells:
        if not cell.text:
            # Eliminar el borde inferior de la celda
            tc = cell._element
            tcPr = tc.get_or_add_tcPr()
            bottom_border = tcPr.find(qn('w:bottom'))
            if bottom_border is not None:
                tcPr.remove(bottom_border)

# Insertar la tabla en la última posición
doc.add_paragraph()._element.addprevious(table._element)

# Eliminar la última fila si contiene solo celdas vacías o "<NA>"
if table.rows:
    last_row = table.rows[-1]
    if all(not cell.text or cell.text == "<NA>" for cell in last_row.cells):
        print("Eliminando la última fila con celdas vacías o <NA>...")
        last_row._element.getparent().remove(last_row._element)

# Añadir la oración después de la tabla
doc.add_paragraph(target_text)

# Guardar el documento con la fecha proporcionada
fecha_actualizada = input("Ingresa la fecha en formato AAAA-MM-DD: ")
nombre_salida = output_path.format(fecha=fecha_actualizada)
doc.save(nombre_salida)

print(f"Documento guardado como: {nombre_salida}")
```
